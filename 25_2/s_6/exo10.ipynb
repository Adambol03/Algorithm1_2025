```python
import math
#Problem 2
def bayes_positive(prior, sensitivity, false_positive):
    """
    Computes P(disease | positive test).
    """
    return (sensitivity * prior) / (sensitivity * prior +
                                    false_positive * (1 - prior))


p = bayes_positive(0.001, 0.99, 0.01)
print("Bayes result (should be ~0.09) =", p)
#Problem 3

def entropy(p):
    """
    Entropy of a biased coin with P(Heads)=p.
    """
    if p == 0 or p == 1:
        return 0
    return -p * math.log(p, 2) - (1 - p) * math.log((1 - p), 2)


coins = [0.5, 0.99, 0.01]

for p in coins:
    print(f"Entropy for p={p}:", entropy(p))
```
