```python

#Problem 8
from collections import defaultdict
#Problem 1

def reverse_graph(graph):
    """Берілген бағытталған графтың кері графын қайтарады."""
    rev = {v: [] for v in graph}

    for u in graph:
        for v in graph[u]:
            rev[v].append(u)

    return rev

#Problem 1
def find_scc(graph):

    visited = set()
    order = []
    def dfs1(u):
        visited.add(u)
        for v in graph[u]:
            if v not in visited:
                dfs1(v)
        order.append(u)

    for node in graph:
        if node not in visited:
            dfs1(node)
    reversed_graph = reverse_graph(graph)
    visited.clear()
    components = []

    def dfs2(u, comp):
        visited.add(u)
        comp.append(u)
        for v in reversed_graph[u]:
            if v not in visited:
                dfs2(v, comp)

    for u in reversed(order):
        if u not in visited:
            comp = []
            dfs2(u, comp)
            components.append(comp)

    return components



#Problem 2

def euler_tour(graph):
    g = {u: list(graph[u]) for u in graph}

    start = next(iter(graph))
    stack = [start]
    path = []

    while stack:
        v = stack[-1]
        if g[v]:
            nxt = g[v].pop()
            stack.append(nxt)
        else:
            path.append(stack.pop())

    return path[::-1]
#Problem 3

def topo_sort(graph):
    visited = set()
    order = []

    def dfs(u):
        visited.add(u)
        for v in graph[u]:
            if v not in visited:
                dfs(v)
        order.append(u)

    for u in graph:
        if u not in visited:
            dfs(u)

    return order[::-1]

if __name__ == "__main__":

    #Test1
    print(" Problem 1 ")
    G1 = {
        "A": ["B"],
        "B": ["C"],
        "C": ["A", "D"],
        "D": []
    }

    print("Reverse:", reverse_graph(G1))
    print("SCC:", find_scc(G1))


    #Test2
    print("\n Problem 2 ")
    G2 = {
        "A": ["B"],
        "B": ["C"],
        "C": ["A"]
    }

    print("Euler tour:", euler_tour(G2))

    #Test3
    print("\n Problem 3 ")
    G3 = {
        "A": ["B", "C"],
        "B": ["C", "D"],
        "C": ["E"],
        "D": ["E", "F"],
        "E": [],
        "F": [],
        "G": ["F", "E"]
    }

    print("Topological order:", topo_sort(G3))
```
