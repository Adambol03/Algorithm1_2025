```python
# Problem Set 9
from itertools import product
#Problem 1
def count_functions_binary(n):
    """Number of functions f: {0,1}^n -> {0,1}"""
    return 2 ** (2 ** n)

def count_functions_ternary(n):
    """Number of functions f: {0,1}^n -> {-1,0,1}"""
    return 3 ** (2 ** n)

def count_functions_mbit(n, m):
    """Number of functions f: {0,1}^n -> {0,1}^m"""
    return (2 ** m) ** (2 ** n)
#Problem 3
def delta(x, y):
    """Implements δ_x(y): returns 1 if y==x, else 0."""
    return 1 if x == y else 0
# Testing
if __name__ == "__main__":
    print(" Problem 1 ")
    n = 3
    m = 2
    print("2^{2^n} =", count_functions_binary(n))
    print("3^{2^n} =", count_functions_ternary(n))
    print("2^{m*2^n} =", count_functions_mbit(n, m))

    print("\n Problem 3 ")
    all_inputs = list(product([0,1], repeat=3))

    x = (1,0,1)
    for y in all_inputs:
        print(f"delta({x},{y}) =", delta(x,y))
```
